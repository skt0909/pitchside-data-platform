```python
from pyspark.sql import functions as F

CATALOG = "workspace"
SCHEMA = "pitchside"
LANDING = f"/Volumes/{CATALOG}/{SCHEMA}/landing"
TABLES = ["teams", "players", "fixtures", "player_gw_stats", "ml_predictions"]

spark.sql(f"USE CATALOG {CATALOG}")
spark.sql(f"USE SCHEMA {SCHEMA}")

for t in TABLES:
    df = spark.read.parquet(f"{LANDING}/{t}.parquet")
    df = df.withColumn("_loaded_at", F.current_timestamp())
    df.write.mode("overwrite").format("delta").saveAsTable(f"raw_{t}")
    print(f"{t}: read {df.count()} rows -> raw_{t}")
```
